```python
#import libraries
import numpy as np
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
#checking null values
df.isnull().sum()
```

```python
#filling null vales with median
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#convert column names to lowerchase and snake case 
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ' , '_')
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#creating a new column based on age group
labels = ['young_adult','adult','middle-aged','senior']
df['age_group'] = pd.qcut(df['age'] , q=4 , labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# creating column purchase_frequency_days

frequency_mapping = {
    'Fortnightly':14,
    'Weekly' :7,
    'Monthly' :30,
   ' Quarterly' :90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months' :90,
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head()
```
